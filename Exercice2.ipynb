# Exercice 2 - Hedge CDS

Vendeur de protection sur corporate A, R = 60%, r = 4%

```python
import numpy as np
```

## Question 1

Le spread CDS dans un modele à intensité constante : $s = \lambda (1-R)$

Donc $\lambda = \frac{s}{1-R}$

```python
R = 0.60
spread_5y = 70e-4  # 70 bps
r = 0.04

lam = spread_5y / (1 - R)
print(f"Intensité de défaut : lambda = {lam:.4f} = {lam*100:.2f}%")
```

### Sensibilité au spread

La RPV01 c'est la sensibilité de la valeur du CDS à un mouvement de 1 bp du spread.

$$RPV01 = \sum_{i=1}^{T} e^{-(r+\lambda)t_i}$$

Pour le vendeur de protection (short), si le spread monte il perd de l'argent, donc sensibilité = -RPV01.

```python
# RPV01 5 ans (paiements annuels)
rpv01_5 = sum(np.exp(-(r + lam)*t) for t in range(1, 6))
print(f"RPV01 5Y = {rpv01_5:.4f}")
print(f"Sensibilité du vendeur = -{rpv01_5:.4f} par bp par unité de notionnel")
```

## Question 2 - Spread 7Y

Si l'intensité est constante, le spread est indépendant de la maturité (la courbe est plate). Donc spread 7Y = spread 5Y = 70 bps.

```python
spread_7y = lam * (1 - R)
print(f"Spread 7Y = {spread_7y*1e4:.0f} bps")
print("Avec lambda constant, la term structure est plate.")
```

## Question 3 - Position pour être insensible au shift parallele

On veut que le PnL total soit nul si les deux spreads bougent du meme montant $\Delta s$ :

$$N_5 \cdot RPV01_5 \cdot \Delta s + N_7 \cdot RPV01_7 \cdot \Delta s = 0$$

Donc : $\frac{N_7}{N_5} = - \frac{RPV01_5}{RPV01_7}$

```python
rpv01_7 = sum(np.exp(-(r + lam)*t) for t in range(1, 8))
print(f"RPV01 7Y = {rpv01_7:.4f}")

ratio = rpv01_5 / rpv01_7
print(f"\nRatio = RPV01_5/RPV01_7 = {ratio:.4f}")
print(f"Le trader doit acheter protection 7Y pour un notionnel = {ratio:.4f} x notionnel short 5Y")
print(f"(signe opposé à la position 5Y)")
```

## Question 4 - P&L

Position : vendeur protection 5Y, nominal 15M€, maturité 5Y.
- Spread 5Y baisse de 3 bps
- Spread 7Y baisse de 1 bps

```python
N = 15_000_000
delta_s5 = -3e-4  # -3 bps

# pour le vendeur, quand le spread baisse c'est un gain
# PnL = -N * RPV01 * delta_s  (signe - car short protection)
pnl = -N * rpv01_5 * delta_s5

print(f"P&L = {pnl:,.0f} €")
print(f"\nC'est un gain : le spread a baissé et on est vendeur de protection (short risque).")
```
