# Exercice 2 - Hedge CDS

Vendeur de protection sur corporate A, R = 60%, r = 4%

```python
import numpy as np
```

## Question 1 - Intensité de défaut et sensibilité

Avec un modele à intensité constante, le spread c'est lambda * (1-R). On en déduit lambda.

```python
R = 0.60
spread_5y = 70e-4  # 70 bps
r = 0.04

lam = spread_5y / (1 - R)
print(f"lambda = {lam:.4f} = {lam*100:.2f}%")
```

La sensibilité au spread c'est la RPV01 : c'est la somme des facteurs d'actualisation risqués (on actualise avec r + lambda pour prendre en compte la survie).

```python
rpv01_5 = sum(np.exp(-(r + lam)*t) for t in range(1, 6))
print(f"RPV01 5Y = {rpv01_5:.4f}")
print(f"Sensibilité du short = -{rpv01_5:.4f} (perd quand spread monte)")
```

## Question 2 - Spread 7Y

Si lambda est constant, le spread ne dépend pas de la maturité (courbe plate). Donc spread 7Y = 70 bps aussi.

```python
spread_7y = lam * (1 - R)
print(f"Spread 7Y = {spread_7y*1e4:.0f} bps (= spread 5Y, normal avec lambda constant)")
```

## Question 3 - Hedge du shift parallèle

Pour être insensible à un shift parallele, il faut que les RPV01 se compensent entre la position 5Y et la position 7Y. Le ratio c'est RPV01_5 / RPV01_7.

```python
rpv01_7 = sum(np.exp(-(r + lam)*t) for t in range(1, 8))
print(f"RPV01 7Y = {rpv01_7:.4f}")

ratio = rpv01_5 / rpv01_7
print(f"\nIl faut acheter protection 7Y pour {ratio:.4f} x le notionnel de la pos 5Y")
```

## Question 4 - P&L

Nominal 15M€, vendeur 5Y, spread 5Y baisse de 3 bps.

```python
N = 15_000_000
delta_s = -3e-4  # baisse de 3 bps

# vendeur de protection : gagne quand le spread baisse
pnl = -N * rpv01_5 * delta_s

print(f"P&L = +{pnl:,.0f} €")
print("C'est un gain car on est short protection et le spread a baissé.")
```
